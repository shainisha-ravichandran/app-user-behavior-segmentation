```python
clustering_features = [
    'sessions_per_week',
    'avg_session_duration_min',
    'daily_active_minutes',
    'feature_clicks_per_session',
    'notifications_opened_per_week',
    'in_app_search_count',
    'pages_viewed_per_session',
    'crash_events_last_30_days',
    'support_tickets_raised',
    'days_since_last_login',
    'ads_clicked_last_30_days',
    'content_downloads',
    'social_shares',
    'churn_risk_score',
    'engagement_score'
]
```

```python
clustering_df = df[clustering_features].copy()
```

```python
clustering_df.head()
```

```python
clustering_df.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(clustering_df)
```

```python
scaled_df = pd.DataFrame(
    scaled_data,
    columns=clustering_features
)
```

```python
scaled_df.head()
```

```python
scaled_df.describe().T[['mean', 'std']]
```

```python
from sklearn.cluster import KMeans

inertia = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(K_range, inertia, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(scaled_data)
    
    score = silhouette_score(scaled_data, labels)
    silhouette_scores.append(score)

for k, score in zip(K_range, silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

# Random sample of 5000 users
sample_data = scaled_data[np.random.RandomState(42).choice(
    scaled_data.shape[0], 5000, replace=False
)]

silhouette_scores = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(sample_data)
    
    score = silhouette_score(sample_data, labels)
    silhouette_scores.append(score)

for k, score in zip(K_range, silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
for k in [3, 4]:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(scaled_data)
    
    print(f"\nK = {k}")
    print("Cluster sizes:")
    print(pd.Series(labels).value_counts().sort_index())
```

```python
final_kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

cluster_labels = final_kmeans.fit_predict(scaled_data)
```

```python
df['cluster'] = cluster_labels
```

```python
engagement_features = [
    'sessions_per_week',
    'avg_session_duration_min',
    'daily_active_minutes',
    'feature_clicks_per_session',
    'notifications_opened_per_week',
    'in_app_search_count',
    'pages_viewed_per_session',
    'content_downloads',
    'social_shares',
    'days_since_last_login',
    'engagement_score',
    'churn_risk_score'
]

engagement_df = df[engagement_features].copy()

engagement_df.shape
```

```python
engagement_scaler = StandardScaler()

engagement_scaled = engagement_scaler.fit_transform(engagement_df)
```

```python
engagement_scaled_df = pd.DataFrame(
    engagement_scaled,
    columns=engagement_features
)

engagement_scaled_df.describe().T[['mean', 'std']]
```

```python
engagement_inertia = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(engagement_scaled)
    engagement_inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(K_range, engagement_inertia, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method - Engagement Features')

plt.show()
```

```python
engagement_sample = engagement_scaled[
    np.random.RandomState(42).choice(
        engagement_scaled.shape[0],
        5000,
        replace=False
    )
]

engagement_silhouette = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(engagement_sample)
    
    score = silhouette_score(
        engagement_sample,
        labels
    )
    
    engagement_silhouette.append(score)

for k, score in zip(range(2, 11), engagement_silhouette):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
final_kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

final_labels = final_kmeans.fit_predict(engagement_scaled)
```

```python
df['cluster'] = final_labels
```
